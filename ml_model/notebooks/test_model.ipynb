```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
```

```python
model_path = "../model"
# 2. تحميل الموديل والمترجم
print("🔄 جاري تحميل الموديل...")
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForSequenceClassification.from_pretrained(model_path)
print("✅ الموديل جاهز!")
```

```python
texts = [
    "A huge earthquake just hit the city, buildings are collapsing!", # كارثة
    "I am going to watch a football match tonight with my friends."    # عادية
]
```

```python
inputs = tokenizer(texts, padding=True, truncation=True, return_tensors="pt")
with torch.no_grad():
    outputs = model(**inputs)
    predictions = torch.nn.functional.softmax(outputs.logits, dim=-1)
```

```python
for text, pred in zip(texts, predictions):
    # افتراضاً: الـ index 1 هو الكارثة، والـ index 0 هو العادي
    is_crisis_prob = pred[1].item() * 100
    print(f"\nText: {text}")
    print(f"Crisis Probability: {is_crisis_prob:.2f}%")
```
